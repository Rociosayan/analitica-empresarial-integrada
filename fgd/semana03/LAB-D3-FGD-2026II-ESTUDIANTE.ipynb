### Cómo se relacionan las tablas

- Un cliente puede tener varios pedidos, y cada pedido pertenece a un solo cliente. Se enlazan por id_cliente: en clientes es la clave primaria y en pedidos es la clave foránea.

- Un pedido puede incluir varias líneas de detalle, y cada línea pertenece a un solo pedido. Se enlazan por id_pedido.

- Un producto puede aparecer en muchas líneas de detalle, y cada línea se refiere a un solo producto. Se enlazan por id_producto.

- Por eso detalles_pedido une a pedidos con productos: un pedido contiene varios productos y un producto se vende en varios pedidos.

- En esta base hay clientes que aún no tienen pedidos y productos que aún no se han vendido; ninguna fila de pedidos o de detalles apunta a un cliente, pedido o producto inexistente.

```python
# Celda 1 de 30
import sqlite3
import random
import pandas as pd

# Base relacional nueva y simulada de TecnoAndina Retail: pedidos del 1 de
# agosto al 30 de septiembre de 2026. No es el conjunto de 72 ventas de la
# Semana 2.
conexion = sqlite3.connect(":memory:")
conexion.execute("PRAGMA foreign_keys = ON")      # SQLite exige activarlo
conexion.executescript("""
CREATE TABLE clientes (
    id_cliente INTEGER PRIMARY KEY,
    nombre     TEXT NOT NULL,
    ciudad     TEXT NOT NULL
);

CREATE TABLE productos (
    id_producto     INTEGER PRIMARY KEY,
    nombre_producto TEXT NOT NULL,
    categoria       TEXT NOT NULL,
    precio_lista    REAL NOT NULL
);

CREATE TABLE pedidos (
    id_pedido    INTEGER PRIMARY KEY,
    id_cliente   INTEGER NOT NULL,
    fecha_pedido TEXT NOT NULL,
    canal        TEXT NOT NULL,
    total        REAL NOT NULL,
    FOREIGN KEY (id_cliente) REFERENCES clientes (id_cliente)
);

CREATE TABLE detalles_pedido (
    id_detalle      INTEGER PRIMARY KEY,
    id_pedido       INTEGER NOT NULL,
    id_producto     INTEGER NOT NULL,
    cantidad        INTEGER NOT NULL,
    precio_unitario REAL NOT NULL,
    FOREIGN KEY (id_pedido)   REFERENCES pedidos (id_pedido),
    FOREIGN KEY (id_producto) REFERENCES productos (id_producto)
);
""")

clientes = [
    (1, "Lucía Paredes", "Lima"), (2, "Carlos Mendoza", "Lima"),
    (3, "Rosa Quispe", "Lima"), (4, "Jorge Salazar", "Lima"),
    (5, "Mónica Rivas", "Lima"), (6, "Andrés Villanueva", "Lima"),
    (7, "Carlos Mendoza", "Arequipa"), (8, "Elena Cáceres", "Arequipa"),
    (9, "Miguel Huamán", "Arequipa"), (10, "Patricia Zevallos", "Arequipa"),
    (11, "Raúl Benavides", "Arequipa"), (12, "Sofía Linares", "Trujillo"),
    (13, "Diego Alva", "Trujillo"), (14, "Valeria Chávez", "Trujillo"),
    (15, "Héctor Burga", "Trujillo"), (16, "Camila Neyra", "Trujillo"),
    (17, "Luis Távara", "Piura"), (18, "Ana Girón", "Piura"),
    (19, "Pedro Seminario", "Piura"), (20, "Marta Infante", "Piura"),
    (21, "Iván Calderón", "Lima"), (22, "Gabriela Ponce", "Arequipa"),
    (23, "Óscar Ruiz", "Trujillo"), (24, "Natalia Ato", "Piura"),
]
productos = [
    (1, "Laptop 14 pulgadas", "Cómputo", 2400),
    (2, "Monitor 24 pulgadas", "Cómputo", 650),
    (3, "Teclado mecánico", "Cómputo", 180),
    (4, "Mouse inalámbrico", "Cómputo", 55),
    (5, "Impresora multifuncional", "Cómputo", 780),
    (6, "Smartphone gama media", "Celulares", 1200),
    (7, "Smartphone gama alta", "Celulares", 3200),
    (8, "Tablet 10 pulgadas", "Celulares", 900),
    (9, "Audífonos Bluetooth", "Accesorios", 160),
    (10, "Cargador rápido", "Accesorios", 70),
    (11, "Funda protectora", "Accesorios", 35),
    (12, "Memoria USB 64 GB", "Accesorios", 40),
    (13, "Parlante portátil", "Accesorios", 220),
    (14, "Reloj inteligente", "Accesorios", 350),
]
conexion.executemany("INSERT INTO clientes VALUES (?, ?, ?)", clientes)
conexion.executemany("INSERT INTO productos VALUES (?, ?, ?, ?)", productos)

# Pedidos por cliente: (en agosto, en septiembre). Los clientes 6, 11, 15, 20,
# 22 y 24 no registran pedidos. Los productos 5 y 14 no se venden.
plan = {1: (2, 2), 2: (1, 3), 3: (2, 1), 4: (1, 0), 5: (0, 1), 7: (1, 1),
        8: (2, 1), 9: (1, 0), 10: (1, 2), 12: (0, 2), 13: (2, 0), 14: (1, 1),
        16: (2, 1), 17: (2, 2), 18: (0, 1), 19: (1, 1), 21: (1, 1), 23: (1, 1)}
vendibles = [1, 2, 3, 4, 6, 7, 8, 9, 10, 11, 12, 13]
pesos = [2, 4, 5, 6, 2, 1, 3, 6, 6, 8, 7, 4]
precio_lista = {p[0]: p[3] for p in productos}

azar = random.Random(3303)    # semilla fija: todos obtienen los mismos datos
cabeceras = []
for id_cliente, (n_ago, n_sep) in plan.items():
    for mes, cantidad_pedidos, dias in ((8, n_ago, 31), (9, n_sep, 30)):
        for _ in range(cantidad_pedidos):
            fecha = f"2026-{mes:02d}-{azar.randint(1, dias):02d}"
            canal = "Web" if azar.random() < 0.3 else "Tienda"
            cabeceras.append((fecha, id_cliente, canal))
cabeceras.sort()                    # los id_pedido siguen el orden cronológico

id_detalle = 0
for id_pedido, (fecha, id_cliente, canal) in enumerate(cabeceras, start=1):
    elegidos = []
    while len(elegidos) < azar.choice((1, 1, 2, 2, 3)):
        producto = azar.choices(vendibles, weights=pesos)[0]
        if producto not in elegidos:
            elegidos.append(producto)
    detalles, total = [], 0
    for id_producto in elegidos:
        lista = precio_lista[id_producto]
        opciones = (1, 1, 2, 3) if lista < 500 else (1, 1, 1, 2)
        cantidad = azar.choice(opciones)
        precio = round(lista * (1 - azar.choice((0, 0, 0, 0.05, 0.10))))
        id_detalle += 1
        detalles.append(
            (id_detalle, id_pedido, id_producto, cantidad, float(precio)))
        total += cantidad * precio
    conexion.execute("INSERT INTO pedidos VALUES (?, ?, ?, ?, ?)",
                     (id_pedido, id_cliente, fecha, canal, float(total)))
    conexion.executemany(
        "INSERT INTO detalles_pedido VALUES (?, ?, ?, ?, ?)", detalles)

conexion.commit()

def mostrar_consulta(consulta):
    """Ejecuta una consulta SQL y muestra su resultado (los NULL se ven NULL)."""
    resultado = pd.read_sql_query(consulta, conexion).convert_dtypes()
    visible = resultado.astype(object).where(resultado.notna(), "NULL")
    print(visible.to_string(index=False, float_format=lambda x: f"{x:.2f}"))
    print(f"Filas: {len(resultado)}")
    return resultado

print("Base preparada: 4 tablas creadas y cargadas con datos simulados.")
```

```python
# Celda 5 de 30
consulta = """SELECT p.id_pedido, p.fecha_pedido,
       c.id_cliente, c.nombre, c.ciudad, p.total
FROM pedidos AS p
INNER JOIN clientes AS c ON p.id_cliente = c.id_cliente
ORDER BY p.fecha_pedido DESC, p.id_pedido DESC
LIMIT 10;"""
mostrar_consulta(consulta)
```

La jefatura quiere comparar a cada cliente entre agosto y septiembre. Hay clientes que compraron solo en uno de los dos meses y ningún resumen debe perderlos. **Se solicita.** Ejecute las tres celdas: la primera crea los dos resúmenes; la segunda los cruza con FULL OUTER JOIN (si la versión lo admite) y la tercera obtiene lo mismo con LEFT JOIN y UNION ALL, que apila los resultados de dos consultas con las mismas columnas.

**Celda 12 de 30**

```python
# Celda 12 de 30
conexion.executescript("""
DROP VIEW IF EXISTS resumen_agosto;
DROP VIEW IF EXISTS resumen_septiembre;
CREATE TEMP VIEW resumen_agosto AS
SELECT id_cliente, COUNT(id_pedido) AS num_pedidos, SUM(total) AS monto
FROM pedidos
WHERE fecha_pedido BETWEEN '2026-08-01' AND '2026-08-31'
GROUP BY id_cliente;
CREATE TEMP VIEW resumen_septiembre AS
SELECT id_cliente, COUNT(id_pedido) AS num_pedidos, SUM(total) AS monto
FROM pedidos
WHERE fecha_pedido BETWEEN '2026-09-01' AND '2026-09-30'
GROUP BY id_cliente;
""")
consulta = """SELECT (SELECT COUNT(*) FROM resumen_agosto) AS clientes_agosto,
       (SELECT COUNT(*) FROM resumen_septiembre) AS clientes_septiembre;"""
mostrar_consulta(consulta)
```

**Celda 13 de 30**

```python
# Celda 13 de 30
consulta = """SELECT COALESCE(a.id_cliente, s.id_cliente) AS id_cliente,
       a.num_pedidos AS pedidos_agosto, a.monto AS monto_agosto,
       s.num_pedidos AS pedidos_septiembre, s.monto AS monto_septiembre
FROM resumen_agosto AS a
FULL OUTER JOIN resumen_septiembre AS s ON a.id_cliente = s.id_cliente
ORDER BY id_cliente;"""
if admite_right_full:
    resultado_full = mostrar_consulta(consulta)
else:
    print("FULL OUTER JOIN no disponible: pase a la celda 14.")
```

**Celda 14 de 30**

```python
# Celda 14 de 30
consulta = """SELECT a.id_cliente,
       a.num_pedidos AS pedidos_agosto, a.monto AS monto_agosto,
       s.num_pedidos AS pedidos_septiembre, s.monto AS monto_septiembre
FROM resumen_agosto AS a
LEFT JOIN resumen_septiembre AS s ON a.id_cliente = s.id_cliente
UNION ALL
SELECT s.id_cliente, NULL, NULL, s.num_pedidos, s.monto
FROM resumen_septiembre AS s
LEFT JOIN resumen_agosto AS a ON s.id_cliente = a.id_cliente
WHERE a.id_cliente IS NULL
ORDER BY 1;"""
resultado_alterno = mostrar_consulta(consulta)
if admite_right_full:
    igual = resultado_alterno.equals(resultado_full)
    print("¿Mismo resultado que FULL OUTER JOIN?", igual)
```

Una subconsulta calcula primero un valor o una tabla intermedia que la consulta principal utiliza después.

**Se solicita.** Ejecute las dos celdas: la primera calcula el promedio de compras acumuladas con las dos definiciones; la segunda lista a los clientes con pedidos cuyo monto acumulado supera el promedio calculado solo entre quienes compraron.

**Celda 20 de 30**

```python
# Celda 20 de 30
consulta = """SELECT
  ROUND(AVG(monto_cliente), 2) AS promedio_solo_con_pedidos,
  ROUND(SUM(monto_cliente) * 1.0 / (SELECT COUNT(*) FROM clientes), 2)
      AS promedio_todos_los_clientes
FROM (SELECT id_cliente, SUM(total) AS monto_cliente
      FROM pedidos
      GROUP BY id_cliente) AS compras;"""
mostrar_consulta(consulta)
```

**Celda 21 de 30**

```python
# Celda 21 de 30
consulta = """SELECT c.id_cliente, c.nombre, c.ciudad,
       COUNT(p.id_pedido) AS num_pedidos,
       SUM(p.total) AS monto_acumulado
FROM clientes AS c
INNER JOIN pedidos AS p ON c.id_cliente = p.id_cliente
GROUP BY c.id_cliente, c.nombre, c.ciudad
HAVING SUM(p.total) > (SELECT AVG(monto_cliente)
                       FROM (SELECT SUM(total) AS monto_cliente
                             FROM pedidos
                             GROUP BY id_cliente) AS compras)
ORDER BY monto_acumulado DESC, c.id_cliente;"""
mostrar_consulta(consulta)
```

```python
# Celda 23 de 30
conexion.executescript("""
DROP VIEW IF EXISTS vista_resumen_cliente;
DROP VIEW IF EXISTS vista_ventas_producto;
CREATE VIEW vista_resumen_cliente AS
SELECT c.id_cliente, c.nombre, c.ciudad,
       COUNT(p.id_pedido)        AS num_pedidos,
       COALESCE(SUM(p.total), 0) AS monto_acumulado,
       ROUND(AVG(p.total), 2)    AS promedio_pedido
FROM clientes AS c
LEFT JOIN pedidos AS p ON c.id_cliente = p.id_cliente
GROUP BY c.id_cliente, c.nombre, c.ciudad;
CREATE VIEW vista_ventas_producto AS
SELECT pr.id_producto, pr.nombre_producto, pr.categoria,
       COALESCE(SUM(d.cantidad), 0) AS unidades_vendidas,
       COALESCE(SUM(d.cantidad * d.precio_unitario), 0) AS ingreso
FROM productos AS pr
LEFT JOIN detalles_pedido AS d ON pr.id_producto = d.id_producto
GROUP BY pr.id_producto, pr.nombre_producto, pr.categoria;
""")
conexion.commit()
consulta = """SELECT name AS vista
FROM sqlite_master
WHERE type = 'view'
ORDER BY name;"""
mostrar_consulta(consulta)
```
